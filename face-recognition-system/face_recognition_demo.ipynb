# Face Recognition System

Encode known faces from a folder, then recognize them in a photo or a live webcam stream.
The functions live in `face_recognition_system.py`; this notebook shows how to use them.

**Before running:** put photos of people who agreed to be included in the `known_faces/` folder (see `known_faces/README.md`).

## 1. Check the environment
This catches the common *dlib / NumPy mismatch* early (the error `Unsupported image type, must be 8bit gray or RGB image`).

```python
import face_recognition_system as frs

frs.check_environment()
```

## 2. Encode the known faces
Each face becomes a 128-number vector. The result is saved to `encodings.npz`.

```python
encodings, names = frs.encode_known_faces(folder='known_faces', out_file='encodings.npz')
```

## 3. Recognize faces in an image
Change the path to your own test photo. The distance shown is the difference between faces: **lower means more similar**.

```python
results = frs.recognize_image('test_images/test.jpg',
                               tolerance=0.5,
                               output_path='output/recognized.jpg')
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.cvtColor(cv2.imread('output/recognized.jpg'), cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.imshow(img)
plt.axis('off')
plt.show()
```
